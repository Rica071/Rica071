## 4 - Criar sequências para o modelo LSTM

```python
def create_sequences(data, seq_length):
    X, y = [], []
    for i in range(len(data) - seq_length):
        X.append(data[i:i+seq_length])
        y.append(data[i+seq_length])
    return np.array(X), np.array(y)
```

```python
seq_length = 10
X, y = create_sequences(data_SMV['Vazão(m³/s)'].values, seq_length)
```

## 5 - Treinar o modelo LSTM separando TREINO e TESTE

```python
train_size = int(0.8 * len(X))
X_train, X_test = X[:train_size], X[train_size:]
y_train, y_test = y[:train_size], y[train_size:]
```

## 5.1 - Redimensionar para AMOSTRAS, PASSO DE TEMPO e CARACTERÍSTICAS

```python
X_train = X_train.reshape(X_train.shape[0], X_train.shape[1], 1)
X_test = X_test.reshape(X_test.shape[0], X_test.shape[1], 1)
```

## 6 - Criar o modelo LSTM

```python
model = Sequential()
model.add(LSTM(50, activation='relu', input_shape=(seq_length, 1)))
model.add(Dense(1))
model.compile(optimizer='adam', loss='mse', metrics=['mae'])
```

## 6.1 - Treinar o modelo LSTM

```python
history = model.fit(X_train, y_train, epochs=50, batch_size=32,
                    validation_data=(X_test, y_test))
```

## 6.2 - Fazer PREVISÕES

```python
predictions = model.predict(X_test)
```

## 6.3 Desnormalizar os resultados

```python
# Assuming 'X' and 'y' are already defined as normalized data

# Instantiate and fit the scaler to the training data
scaler = MinMaxScaler()
scaler.fit(X_train.reshape(-1, X_train.shape[-1])) # Fit the scaler on the training data

predictions = model.predict(X_test)

# Reshape predictions to 2D before inverse transforming
predictions = predictions.reshape(-1, 1)

predictions = scaler.inverse_transform(predictions)
y_test = scaler.inverse_transform(y_test.reshape(-1, 1))
```

```python
def create_multivariate_sequences(data, target_col, seq_length):
    X, y = [], []
    for i in range(len(data) - seq_length):
        X.append(data.iloc[i:i + seq_length].values)
        y.append(data.iloc[i + seq_length][target_col])
    return np.array(X), np.array(y)
```

```python
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
```

```python
dfBatalha = pd.read_excel("/content/drive/MyDrive/Res_Descarga_Batalha_.xlsx")
dfBatalha
```

# Ajustar nomes de colunas conforme necessário

```python
dfBatalha.columns = dfBatalha.columns.str.strip()
dfBatalha['Data'] = pd.to_datetime(dfBatalha['Data'])
dfBatalha = dfBatalha.sort_values('Data')
```

# Exemplo: coluna de vazão


```python
vazao = dfBatalha['Vazao']
```

# --- 1. Método estatístico clássico: Z-score ---
# valores acima de 3 desvios-padrão são possíveis outliers

```python
z_scores = np.abs(stats.zscore(vazao, nan_policy='omit'))
limite = 3
outliers_z = dfBatalha[z_scores > limite]
```


# --- 2. Método baseado em IQR (Interquartile Range) ---

```python
Q1 = vazao.quantile(0.25)
Q3 = vazao.quantile(0.75)
IQR = Q3 - Q1
limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR
outliers_iqr = dfBatalha[(vazao < limite_inferior) | (vazao > limite_superior)]
```


# --- 3. Visualização ---


```python
plt.figure(figsize=(10, 5))
plt.plot(dfBatalha['Data'], vazao, label='Vazão', color='blue')
plt.scatter(outliers_iqr['Data'], outliers_iqr['Vazao'], color='red', label='Outliers (IQR)')
plt.title("Detecção de Outliers em Vazão - Estação Batalha")
plt.xlabel("Data")
plt.ylabel("Vazão (m³/s)")
plt.legend()
plt.show()

print("Número de outliers (IQR):", len(outliers_iqr))
```
